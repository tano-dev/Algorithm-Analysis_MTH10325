### Bài tập:

$$T_n = \begin{cases} C_1, & n = 1 \\ 3 \cdot T\left(\lfloor \frac{n}{2} \rfloor\right) + C_2 n, & n \ge 2 \end{cases}$$

Chứng minh $$T(n) = O(n^{\log_2 3})$$

<center>
Bài làm
</center>

ta giả sử $n = 2^k$ (với $k \ge 0$). Lúc này, $k = \log_2 n$. Ta có
$$T(n) = 3T\left(\frac{n}{2}\right) + C_2 n$$

Thay $T\left(\frac{n}{2}\right) = 3T\left(\frac{n}{4}\right) + C_2 \frac{n}{2}$ vào, ta được
$$T(n) = 3\left(3T\left(\frac{n}{4}\right) + C_2 \frac{n}{2}\right) + C_2 n = 3^2 T\left(\frac{n}{4}\right) + 3 C_2 \frac{n}{2} + C_2 n$$

Thay tiếp $T\left(\frac{n}{4}\right)$:
$$T(n) = 3^2 \left(3T\left(\frac{n}{8}\right) + C_2 \frac{n}{4}\right) + 3 C_2 \frac{n}{2} + C_2 n = 3^3 T\left(\frac{n}{8}\right) + 3^2 C_2 \frac{n}{4} + 3 C_2 \frac{n}{2} + C_2 n$$

...

Tổng quát sau $k$ lần thay thế sẽ là:
$$T(n) = 3^k T\left(\frac{n}{2^k}\right) + C_2 n \sum_{i=0}^{k-1} 3^i \cdot \frac{1}{2^i}$$

Thuật toán dừng khi $\frac{n}{2^k} = 1$, tức là $k = \log_2 n$. Khi đó, ta có:
$$T(n) = 3^{\log_2 n} T(1) + C_2 n \sum_{i=0}^{\log_2 n - 1} 3^i \cdot \frac{1}{2^i}$$

Mà ta có $3^{\log_2 n} = n^{\log_2 3}$ và công thức tổng cấp số nhân:
$$\sum_{i=0}^{m} r^i = \frac{r^{m+1} - 1}{r - 1}$$
với $r = \frac{3}{2}$ và $m = \log_2 n - 1$, ta có:
$$\sum_{i=0}^{\log_2 n - 1} \left(\frac{3}{2}\right)^i = \frac{\left(\frac{3}{2}\right)^{\log_2 n} - 1}{\frac{3}{2} - 1} = \frac{\left(\frac{3}{2}\right)^{\log_2 n} - 1}{\frac{1}{2}} = 2 \left(\left(\frac{3}{2}\right)^{\log_2 n} - 1\right)$$
Ta có:
$$\left(\frac{3}{2}\right)^{\log_2 n} = n^{\log_2 \frac{3}{2}} = n^{\log_2 3 - 1} = \frac{n^{\log_2 3}}{n}$$
Do đó:
$$\sum_{i=0}^{\log_2 n - 1} \left(\frac{3}{2}\right)^i = 2 \left(\frac{n^{\log_2 3}}{n} - 1\right) = 2 \left(n^{\log_2 3 - 1} - 1\right)$$
Vậy:
$$T(n) = n^{\log_2 3} T(1) + C_2 n \cdot 2 \left(n^{\log_2 3 - 1} - 1\right) = n^{\log_2 3} T(1) + 2 C_2 n^{\log_2 3} - 2 C_2 n$$
Do đó, $T(n) = O(n^{\log_2 3})$.

### Bài tập:

$$T_n = \begin{cases} C_1, & n = 1 \\ 4 \cdot T\left(\lfloor \frac{n}{2} \rfloor\right) + C_2 n, & n \ge 2 \end{cases}$$

Chứng minh $$T(n) = O(n^{2})$$

<center>
Bài làm
</center>

$$T_n = \begin{cases} C_1, & n = 1 \\ 4 \cdot T\left(\lfloor \frac{n}{2} \rfloor\right) + C_2 n, & n \ge 2 \end{cases}$$

Chứng minh $$T(n) = O(n^{2})$$

Ta giả sử $n = 2^k$ (với $k \ge 0$). Lúc này, $k = \log_2 n$. Ta có
$$T(n) = 4T\left(\frac{n}{2}\right) + C_2 n$$

Thay $T\left(\frac{n}{2}\right) = 4T\left(\frac{n}{4}\right) + C_2 \frac{n}{2}$ vào, ta được
$$T(n) = 4\left(4T\left(\frac{n}{4}\right) + C_2 \frac{n}{2}\right) + C_2 n = 4^2 T\left(\frac{n}{4}\right) + 4 C_2 \frac{n}{2} + C_2 n$$

Thay tiếp $T\left(\frac{n}{4}\right)$:
$$T(n) = 4^2 \left(4T\left(\frac{n}{8}\right) + C_2 \frac{n}{4}\right) + 4 C_2 \frac{n}{2} + C_2 n = 4^3 T\left(\frac{n}{8}\right) + 4^2 C_2 \frac{n}{4} + 4 C_2 \frac{n}{2} + C_2 n$$

Tổng quát sau $k$ lần thay thế sẽ là:
$$T(n) = 4^k T\left(\frac{n}{2^k}\right) + C_2 n \sum_{i=0}^{k-1} 4^i \cdot \frac{1}{2^i}$$

Thuật toán dừng khi $\frac{n}{2^k} = 1$, tức là $k = \log_2 n$. Khi đó, ta có:
$$T(n) = 4^{\log_2 n} T(1) + C_2 n \sum_{i=0}^{\log_2 n - 1} 4^i \cdot \frac{1}{2^i}$$

Mà ta có $4^{\log_2 n} = n^{\log_2 4} = n^2$ và công thức tổng cấp số nhân:
$$\sum_{i=0}^{m} r^i = \frac{r^{m+1} - 1}{r - 1}$$
với $r = \frac{4}{2} = 2$ và $m = \log_2 n - 1$, ta có:
$$\sum_{i=0}^{\log_2 n - 1} 2^i = \frac{2^{\log_2 n} - 1}{2 - 1} = n - 1$$
Vậy:
$$T(n) = n^2 T(1) + C_2 n (n - 1) = n^2 T(1) + C_2 n^2 - C_2 n$$
Do đó, $T(n) = O(n^{2})$.
